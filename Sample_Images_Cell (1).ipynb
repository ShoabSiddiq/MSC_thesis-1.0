```python
# --- Show sample images from each class (original, natural-looking) ---
import os
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing import image as kimage

# Path to your training folder (change if needed)
data_dir = '/content/drive/MyDrive/Research/original_data_leveling/up_train'

# The three class subfolders
classes = os.listdir(data_dir)
classes = [c for c in classes if os.path.isdir(os.path.join(data_dir, c))]
classes.sort()
print("Classes found:", classes)

n_per_class = 3   # how many images to show per class

fig, axes = plt.subplots(len(classes), n_per_class, figsize=(4*n_per_class, 4*len(classes)))
for row, cls in enumerate(classes):
    cls_dir = os.path.join(data_dir, cls)
    files = [f for f in os.listdir(cls_dir) if f.lower().endswith(('.jpg','.jpeg','.png'))]
    for col in range(n_per_class):
        img = kimage.load_img(os.path.join(cls_dir, files[col]), target_size=(224,224))
        ax = axes[row, col] if len(classes) > 1 else axes[col]
        ax.imshow(img)
        ax.axis('off')
        if col == 0:
            ax.set_ylabel(cls, fontsize=11)
        ax.set_title(cls if col == n_per_class//2 else "", fontsize=10)
plt.tight_layout()
plt.show()
```
